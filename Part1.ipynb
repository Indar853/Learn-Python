## Variables

## Data Types in Python

Python has several built-in data types:

- `int` — whole numbers like `10`, `-3`
- `float` — decimal numbers like `3.14`, `2.0`
- `str` — text like `"Hello"` or `"John"`
- `bool` — logical values: `True` or `False`
- `list` — ordered, changeable collection: `[1, 2, 3]`
- `tuple` — ordered, unchangeable collection: `(1, 2, 3)`
- `dict` — key-value pairs: `{"name": "John", "age": 25}`

You can check a value’s type using `type()` like `type(course_name)`.

### `int()` Data Type

`int` is a Python data type used to store whole numbers. It can be positive, negative, or zero.

Examples:
- `5`
- `-12`
- `0`

You can assign integers to variables like this:
`age = 25`
`score = 90`
`count = 0`

Python allows you to do arithmetic with integers:

- Addition: `7 + 3 = 10`
- Subtraction: `9 - 4 = 5`
- Multiplication: `6 * 2 = 12`
- Division: `8 / 2 = 4.0` (this produces a float, not an int)
- Modulus: `10 % 3 = 1`

Integers are commonly used for counts, ages, scores, and other numeric values. For example, in this notebook, `duration` is an `int` and its value is `3`.

You can also check the type of a value using `type()`:
`type(duration)` will return `<class 'int'>`

This makes `int` one of the most important and commonly used data types in Python.


````python
# Example of int data type in Python

age = 25
score = 90
total = age + score

print("age:", age)
print("score:", score)
print("total:", total)

print(type(age))      # <class 'int'>
print(type(score))    # <class 'int'>
print(type(total))    # <class 'int'>

# Integer operations
print("Addition:", 10 + 5)
print("Subtraction:", 10 - 5)
print("Multiplication:", 10 * 5)
print("Division:", 10 / 5)  # This gives float, not int
print("Division:", 10 // 5)  # This gives integer, not float
print("Modulo:", 10 % 3)
````

### `float()` Data Type
`float` is a Python data type used to store decimal numbers, also called floating-point numbers. These values can include digits after the decimal point, such as `3.14`, `8.7`, or `0.5`.

Examples:
- `gpa = 8.7`
- `pi = 3.14159`
- `temperature = 25.5`

Python allows you to do arithmetic with floats just like integers:

- Addition: `8.5 + 1.2 = 9.7`
- Subtraction: `10.0 - 2.5 = 7.5`
- Multiplication: `2.5 * 4 = 10.0`
- Division: `9 / 2 = 4.5`

Unlike `int`, which stores whole numbers only, `float` is used for measurements, percentages, averages, and values that require precision beyond whole numbers. In this notebook, `gpa` is a `float` and its value is `8.7`.

You can check the type of a value using `type()`:

```python
print(type(gpa))  # <class 'float'>
```

You can also convert other numeric types to float using `float()`:

```python
score = 90
score_float = float(score)
print(score_float)  # 90.0
```

Floats are very important in scientific calculations, data analysis, and machine learning, where decimal values are common.

````python
# Example of the float data type
gpa = 8.75
print("GPA:", gpa)
print("Type:", type(gpa))

# Float arithmetic
new_gpa = gpa + 0.5
print("GPA after increase:", new_gpa)
````

### Bool Data type
`bool` is a Python data type used to store truth values: `True` and `False`. It is commonly used in comparisons and conditions, such as `age > 18` or `course == "Python Programming"`. Booleans are essential for decision-making in `if` statements and loops, and they work with logical operators like `and`, `or`, and `not`. In Python, `True` and `False` are special values that represent binary logic.

````python
# Bool data type examples

is_student = True
is_adult = age >= 18
has_course = course_name == "Python Programming"
is_python_course = course_name == "Introduction to Python"

print("Type of is_student:", type(is_student))
print("is_student:", is_student)
print("is_adult:", is_adult)
print("has_course:", has_course)
print("is_python_course:", is_python_course)

# Boolean operations
print("AND:", is_student and has_course)
print("OR:", is_student or is_adult)
print("NOT:", not is_student)

# Converting values to bool
print("bool('25'):", bool(course_name))
print("bool(False):", bool(is_student))
print("bool(0):", bool(0))
print("bool(5):", bool(5))
````

# Type Conversion

## Type Conversion in Python

Type conversion is the process of changing one data type into another in Python. It is useful when you want to use values in different operations.

Common conversion functions:
- `int()` → converts to integer
- `float()` → converts to float
- `str()` → converts to string
- `bool()` → converts to boolean

Examples:

````python
x = "10"
y = int(x)      # "10" -> 10
z = float(x)    # "10" -> 10.0
a = str(25)     # 25 -> "25"
b = bool(0)     # 0 -> False
````

Important:
- `int("3.5")` will not work directly; convert via `float()` first.
- `input()` always returns a string, so you often need `int()` or `float()` to convert user input.

Type conversion helps make programs more flexible and easier to work with.

## Type Casting in Python

Type casting is the process of converting a value from one data type to another. Python provides built-in functions for explicit type casting:

- `int()` converts a value to an integer
- `float()` converts a value to a floating-point number
- `str()` converts a value to a string
- `bool()` converts a value to a Boolean

For example, the variable `x` contains the string `"10"`. It can be converted to an integer or a float:

```python
x = "10"

number = int(x)
decimal = float(x)

print(number, type(number))    # 10 <class 'int'>
print(decimal, type(decimal))  # 10.0 <class 'float'>
```

You can also convert numbers to strings:

```python
age_text = str(5)
print(age_text, type(age_text))  # 5 <class 'str'>
```

Some conversions require care. For example, `int("3.5")` causes an error because `"3.5"` is not an integer-formatted string. Convert it to a float first if needed:

```python
value = int(float("3.5"))  # 3
```

Type casting is useful when combining values of different types or processing user input, since `input()` always returns a string.

## Type Conversion vs Type Casting in Python

Both terms mean changing one data type into another, but they are used a little differently.

- **Type conversion** is the general idea of changing a value from one type to another.
    - It can happen automatically or manually.
    - Example: `x = "10"` and then `int(x)` converts it to an integer.
    - Example: `float("3.5")` converts a string to a float.

- **Type casting** is a more specific term for explicit conversion done by the programmer.
    - It is usually done intentionally when a program needs a value in a certain type.
    - Example: `age = 5` then `str(age)` converts it to `"5"`.

### Main difference
- Type conversion is the broader concept.
- Type casting is the explicit, intentional conversion performed by the developer.

### Example
`x = "10"`
`y = int(x)` → converts `"10"` to `10`
`z = str(y)` → converts `10` to `"10"`

### In simple words
- Type conversion = changing data type.
- Type casting = deliberately changing data type using functions like `int()`, `float()`, `str()`, and `bool()`.

Both are very useful in Python, especially when working with user input, calculations, and storing values in the right format.

# Operators

## Operators in Python

An operator is a special symbol or keyword that performs an operation on one or more values called operands and gives a result.

Examples: `+`, `-`, `*`, `/`, `==`, `and`, `not`.

Operators are used for arithmetic, comparison, assignment, logic, and decision-making in Python.

## Types of Operators in Python

- **Arithmetic Operators**: used for mathematical calculations  
    Example: `+`, `-`, `*`, `/`, `%`, `//`, `**`

- **Comparison Operators**: used to compare two values  
    Example: `==`, `!=`, `>`, `<`, `>=`, `<=`

- **Assignment Operators**: used to assign values to variables  
    Example: `=`, `+=`, `-=`, `*=`, `/=`

- **Logical Operators**: used to combine conditions  
    Example: `and`, `or`, `not`

- **Bitwise Operators**: work on binary values  
    Example: `&`, `|`, `^`, `~`, `<<`, `>>`

- **Identity Operators**: check whether two objects are the same  
    Example: `is`, `is not`

- **Membership Operators**: check whether a value exists in a sequence  
    Example: `in`, `not in`

- **Unary Operators**: operate on a single value  
    Example: `-x`, `+x`, `not x`

Example:
`a = 10`
`b = 5`
`print(a + b)` → `15`
`print(a > b)` → `True`
`print(a and b)` → `5`

Operators are fundamental in Python because they help with calculations, conditions, loops, and data processing.

## Arithemetic operator

````python
print(5 + 3) # addition operation
print(5 - 3) # subtraction operation
print(5 * 3) # multiplication operation
print(5 / 3) # division operation
print(5 // 3) # floor division operation gives the quotient of the division
print(5 % 3) # modulo operation gives the remainder of the division

# Operators precedence
# Multiplication and division are performed before addition and subtraction
print(2 + 3 * 4) # multiplication is performed first
print((2 + 3) * 4) # parentheses override the default precedence
````

## Comparision Operators

````python
print(3 > 2) # Greater than operator, returns True
print(3 < 2) # Less than operator, returns False
print(3 == 2) # Equal to operator, returns False
print(3 != 2) # Not equal to operator, returns True

````

## Logical Operators

````python
statement1 = 3 > 5
statement2 = 3 > 2
print(statement1 or statement2) # Logical OR operator, returns True
print(statement1 and statement2) # Logical AND operator, returns False

print(not statement1) # Logical NOT operator, returns True

````

## Lists in Python

A list is a built-in data type used to store multiple items in a single variable. Lists are:

- ordered
- mutable (changeable)
- allow duplicate values
- indexed starting from 0

### Creating a list

````python
empty_list = []
numbers = [10, 20, 30, 40]
names = ["Alice", "Bob", "Charlie"]
mixed = [1, "Hello", True, 3.14]
````

### Accessing list items

````python
print(numbers[0])      # 10
print(names[1])       # Bob
print(mixed[2])       # True


# You can also use slicing:


print(numbers[1:3])   # [20, 30]
print(numbers[:2])    # [10, 20]
print(numbers[-1])    # 40
````

### Changing list items


````python
numbers[1] = 25
print(numbers)         # [10, 25, 30, 40]
````

### Common list methods

````python
fruits = ["apple", "banana"]

fruits.append("orange")      # add at the end
fruits.insert(1, "mango")    # add at a specific index
fruits.remove("banana")      # remove item
fruits.pop()                 # remove last item
fruits.sort()                # sort in ascending order
fruits.reverse()             # reverse the list
print(len(fruits))          # length of the list
print(fruits)
````

### List operations

````python
nums = [1, 2, 3]
nums2 = [4, 5]

total = nums + nums2
print(total)        # [1, 2, 3, 4, 5]

print(2 in nums)    # True
print(len(nums))    # 3
````

- Lists can hold different data types at the same time.

Lists are one of the most commonly used data structures in Python, especially for data storage, loops, and processing collections of values.

````python
#  with for loop 
for student in students:
    print(student)
````

## Set in Python

A set is an unordered, mutable collection of unique elements in Python. It is commonly used when you want to store values without duplicates and perform fast membership checks.

### Characteristics of a Set

- Unordered: items do not have a defined index
- Mutable: you can add or remove elements
- Unique: duplicate values are automatically removed
- Can contain different data types, but usually similar values
- Supports mathematical operations like union, intersection, and difference

### Creating a Set

````python
# Empty set
empty_set = set()

# Set with values
numbers = {1, 2, 3, 4, 5}
names = {"Alice", "Bob", "Charlie"}

# Duplicate values are removed automatically
unique_numbers = {1, 2, 2, 3, 3, 4}
print(unique_numbers)   # {1, 2, 3, 4}
````

### Accessing Set Items

Sets do not support indexing like lists because they are unordered. Instead, you usually check membership using `in`.

````python
numbers = {10, 20, 30, 40}

print(20 in numbers)   # True
print(50 in numbers)   # False
````

### Adding and Removing Elements

````python
fruits = {"apple", "banana"}

fruits.add("orange")       # add one item
fruits.update(["mango", "grape"])   # add multiple items

fruits.remove("banana")    # removes item, raises error if missing
fruits.discard("pear")     # removes item safely if present

print(fruits)
````

### Common Set Methods

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a.union(set_b))      # {1, 2, 3, 4, 5, 6}
print(set_a.intersection(set_b)) # {3, 4}
print(set_a.difference(set_b))   # {1, 2}
print(set_a.symmetric_difference(set_b))  # {1, 2, 5, 6}
````

### Set Operations

- `union()` → combine two sets
- `intersection()` → common elements
- `difference()` → elements in one set but not the other
- `symmetric_difference()` → elements in either set, but not both

### Important Note

Since sets are unordered, you cannot access elements using an index like `my_set[0]`. If you need order, use a list or tuple instead.

### Example

````python
students = {"Hritik", "Aman", "Riya", "Hritik"}

print(students)   # {'Hritik', 'Aman', 'Riya'}

# Here, `"Hritik"` appears only once because sets do not allow duplicate values.
````

### Summary

A set is useful when:

- you need unique values
- you want quick membership checking
- you want to perform mathematical set operations

Sets are one of the most important built-in data structures in Python, especially in tasks involving data filtering, uniqueness, and comparisons.

## Dictionary in Python

A dictionary is a built-in data type in Python that stores data in key-value pairs. Each key is unique and is used to access its corresponding value.

### Why use a dictionary?
Dictionaries are useful when you want to store related information in a structured way, like:

- student names and their marks
- product names and prices
- employee IDs and names

### Creating a dictionary

````python
student = {
    "name": "Indra",
    "age": 25,
    "course": "Python Programming"
}
````

Here:

- `"name"` is the key
- `"Indra"` is the value
- `"age"` is another key
- `25` is the corresponding value

### Accessing values

You can access the value of a key using square brackets:

````python
print(student["name"])   # Indra
print(student["age"])    # 25
````

You can also use the `get()` method:

````python
print(student.get("course"))   # Python Programming
````

### Adding or updating values

````python
student["city"] = "Delhi"
student["age"] = 26

print(student)
````

This adds a new key-value pair and updates the age.

### Removing items

````python
student.pop("city")         # removes the key "city"
del student["age"]          # deletes the key "age"

print(student)
````

### Common dictionary methods

````python
student = {
    "name": "Indra",
    "age": 25,
    "course": "Python Programming"
}

print(student.keys())       # returns all keys
print(student.values())     # returns all values
print(student.items())      # returns key-value pairs
print(len(student))         # number of items
````

### Looping through a dictionary

````python
for key in student:
    print(key, ":", student[key])
````

You can also loop through keys and values separately:

````python
for key, value in student.items():
    print(key, ":", value)
````

### Nested dictionaries

A dictionary can contain another dictionary as a value:

````python
students = {
    "student1": {
        "name": "Aman",
        "age": 20
    },
    "student2": {
        "name": "Riya",
        "age": 22
    }
}

print(students["student1"]["name"])   # Aman
````

### Dictionary vs list

- List stores data in an ordered sequence.
- Dictionary stores data in key-value pairs.
- Lists use index numbers like `0, 1, 2`.
- Dictionaries use meaningful keys like `"name"`, `"age"`, `"course"`.

### Example program

````python
employee = {
    "name": "Ritesh",
    "department": "IT",
    "salary": 50000
}

print("Employee Name:", employee["name"])
print("Department:", employee["department"])
print("Salary:", employee["salary"])

employee["salary"] = 55000
print("Updated Salary:", employee["salary"])
````

A dictionary is one of the most important data structures in Python because it helps organize data efficiently using keys and values. It is especially useful when working with real-world data such as records, settings, configurations, and user information.

Dictionaries are:

- unordered (before Python 3.7, though insertion order is preserved in modern Python)
- mutable
- indexed by keys instead of numbers
- ideal for storing related data in pairs

In simple words, a dictionary is like a real-world phonebook or database record where each item has a label and a value.

# Functions in Python

A function is a reusable block of code that performs a specific task. Functions help us organize code, avoid repetition, and make programs easier to read and maintain.



In Python, a function is defined using the `def` keyword.

````python
def greet():
    print("Hello, World!")

greet()
````

## 1. Why use functions?

Functions are useful because they:

- reduce repetition
- make code cleaner
- make programs easier to debug
- help divide a big problem into smaller parts
- allow code reuse

Example:

````python
def add_numbers():
    a = 10
    b = 20
    print(a + b)

add_numbers()
````

## 2. Function syntax

````python
def function_name(parameters):
    # body of the function
    # statements
    return value
````

- `def` → keyword used to define a function
- `function_name` → name of the function
- `parameters` → values passed to the function
- `return` → optional value returned by the function

## 3. Function with parameters

A function can accept input values called parameters.

````python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Hritik")
greet_user("Riya")
````

## 4. Return values

A function can return a result using `return`.

````python
def add(a, b):
    return a + b

result = add(5, 7)
print(result)
````

If a function does not use `return`, it returns `None` by default.


````python
def print_message():
    print("This prints but does not return any value")

x = print_message()
print(x)
````

## 5. Multiple parameters

````python
def student_info(name, age, course):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Course: {course}")

student_info("Aman", 20, "Python")
````

## 6. Default parameters

You can assign default values to parameters.

````python
def greet_user(name="Guest"):
    print(f"Hello, {name}!")

greet_user()
greet_user("Riya")
````

## 7. Keyword arguments

Arguments can be passed by name.

````python
def student_details(name, course):
    print(f"{name} is studying {course}")

student_details(name="Hritik", course="Python")
student_details(course="Java", name="Aman")
````

## 8. Arbitrary arguments (*args)

If you do not know how many arguments will be passed, use `*args`.

````python
def add_all(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(add_all(1, 2, 3))
print(add_all(10, 20, 30, 40))
````

## 9. Arbitrary keyword arguments (**kwargs)

If you want to pass many named arguments, use `**kwargs`.

````python
def display_info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

display_info(name="Riya", age=22, course="Python")
````

## 10. Scope of variables

Variables defined inside a function are local to that function.

````python
def test():
    x = 10
    print("Inside function:", x)

test()
# print(x)  # This will cause an error because x is local to test()
````

A variable defined outside a function is global.

````python
x = 50

def show():
    print("Global variable:", x)

show()
````

You can use `global` to modify a global variable inside a function:

````python
count = 0

def increase():
    global count
    count += 1

increase()
print(count)
````

## 11. Docstrings

Docstrings are used to explain what a function does.

````python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

print(add(4, 5))
print(add.__doc__)
````

## 12. Lambda functions

A lambda function is a small anonymous function created in one line.

````python
square = lambda x: x * x
print(square(5))
````

Lambda functions are useful for short operations.


````python
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x * x, numbers))
print(squared)
````

## 13. Recursion

A recursive function calls itself.

````python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
````

Important:
- recursion must have a stopping condition
- otherwise it will continue forever

## 14. Nested functions

A function can be defined inside another function.

````python
def outer():
    print("Outer function")

    def inner():
        print("Inner function")

    inner()

outer()
````

## 15. Function returning multiple values

A function can return multiple values as a tuple.

````python
def calculate(a, b):
    addition = a + b
    subtraction = a - b
    return addition, subtraction

result = calculate(10, 5)
print(result)
print(result[0], result[1])
````

Python already provides many built-in functions, such as:

````python
print(len("Python"))
print(type(10))
print(max(3, 7, 1))
print(min(3, 7, 1))
print(abs(-10))
print(round(3.756, 2))
````

## 17. Function example: check prime number

````python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

numbers = [2, 3, 4, 5, 7, 11, 12]

for n in numbers:
    if is_prime(n):
        print(f"{n} is prime")
````

## 18. Function example: calculator

````python
def calculator(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return a / b
    else:
        return "Invalid operation"

print(calculator(10, 5, "add"))
print(calculator(10, 5, "subtract"))
print(calculator(10, 5, "multiply"))
print(calculator(10, 5, "divide"))
````

## 19. Best practices for functions

- use descriptive names
- keep functions small and focused
- write docstrings
- avoid too many parameters
- return values instead of printing when possible
- use default values carefully
- avoid global variables if possible

Example of a well-structured function:

````python
def calculate_area(length, width):
    """Return the area of a rectangle."""
    return length * width

area = calculate_area(6, 4)
print(area)
````
