# Advanced Exception Handling in Python

Exception handling lets you respond to errors without abruptly terminating a program. Advanced techniques help make error handling precise, informative, and maintainable.

## 1. Catch Specific Exceptions

Catch the exceptions you expect and can handle. Avoid a bare `except`, which can hide bugs and intercept unexpected errors.

```python
try:
    value = int("not a number")
except ValueError as error:
    print(f"Invalid integer: {error}")
```

Use separate handlers when different errors require different responses:

```python
divisor = 0  # Change this value to test different scenarios

try:
    result = 10 / divisor
except ZeroDivisionError:
    print("Divisor cannot be zero.")
except TypeError:
    print("Divisor must be a number.")
```

## 2. Use `else` and `finally`

The `else` block runs only when the `try` block completes without an exception. The `finally` block runs whether an exception occurred or not.

```python
divisor = 0  # Change this value to test different scenarios
try:
    result = 10 / divisor
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print(f"Result: {result}")
finally:
    print("Calculation finished.")
```

Use `finally` for cleanup that must happen in either case. For files and other resources, a context manager is often preferable.


## 3. Raise Exceptions Intentionally

Use `raise` to signal that a function cannot complete its task or that its input is invalid.

```python
def set_temperature(value):
    if value < -273.15:
        raise ValueError("Temperature cannot be below absolute zero.")
    return value
```

Validate inputs at the boundary of your program, where untrusted data enters.

## 4. Create Custom Exceptions

Custom exceptions make application errors easier to identify and handle. Derive them from `Exception`.

```python
class PaymentError(Exception):
    """Base exception for payment-related errors."""

class InsufficientFundsError(PaymentError):
    """Raised when an account has insufficient funds."""

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError("Not enough funds.")
    return balance - amount
```

A caller can handle a specific error or the broader category:

```python
try:
    balance = withdraw(50, 75)
except InsufficientFundsError as error:
    print(error)
except PaymentError as error:
    print(f"Payment failed: {error}")
```

## 5. Preserve Exception Context

When translating an exception into a more meaningful one, use `raise ... from ...` to preserve the original cause.

```python
def load_settings(path):
    try:
        with open(path, encoding="utf-8") as file:
            return file.read()
    except OSError as error:
        raise RuntimeError(f"Could not load settings from {path}") from error
```

Use `raise NewError(...) from None` only when intentionally suppressing the displayed context.

## 6. Define Exception Boundaries

Catch exceptions where the program can meaningfully recover, report the issue, or add useful context. Avoid catching an exception just to ignore it.

```python

from matplotlib.pylab import record

def process_record(record):
    if not isinstance(record, dict):
        raise ValueError("Record must be a dictionary.")
    # Process the record (placeholder for actual processing logic)
    print(f"Processing record: {record}")

def log_invalid_record(record, error):
    print(f"Invalid record: {record}. Error: {error}")

try:
    process_record(record)
except ValueError as error:
    log_invalid_record(record, error)
```

Do not normally catch `BaseException`; it includes control-flow exceptions such as `KeyboardInterrupt` and `SystemExit`. Catch `Exception` or a more specific exception when appropriate.

## 7. Use Context Managers for Cleanup

A `with` statement ensures that resources are released even when an exception occurs.

```python
# with open("data.txt", encoding="utf-8") as file:
#     contents = file.read()
```

Context managers are commonly used for files, locks, database transactions, and other resources that need reliable cleanup.

## 8. Handle Multiple Exceptions with `ExceptionGroup`

Python 3.11 introduced `ExceptionGroup` and `except*` for handling multiple exceptions raised together, such as during concurrent tasks.

```python
try:
    raise ExceptionGroup(
        "Several operations failed",
        [ValueError("Invalid value"), TypeError("Invalid type")],
    )
except* ValueError as errors:
    print(f"Value errors: {len(errors.exceptions)}")
except* TypeError as errors:
    print(f"Type errors: {len(errors.exceptions)}")
```

Each `except*` clause handles matching exceptions from the group. This feature is especially useful with structured concurrency.

## Best Practices

- Catch specific exceptions and handle them meaningfully.
- Keep `try` blocks as small as practical.
- Use custom exceptions to represent errors specific to your application.
- Preserve the original cause when re-raising with additional context.
- Use context managers for resource cleanup.
- Avoid silently ignoring exceptions or using exceptions for ordinary control flow.
- Do not catch `BaseException` unless you have a specific reason.

## Summary

Advanced exception handling combines specific handlers, custom exception types, exception chaining, cleanup, and clear recovery boundaries. These practices make failures easier to diagnose while keeping programs predictable and maintainable.

# OOP in Python

Python supports object-oriented programming (OOP), where data and behavior are grouped into classes and objects. At an advanced level, OOP is not just about creating classes — it is about designing flexible, reusable, and maintainable systems.

## 1. Classes and Objects
A class is a blueprint, and an object is an instance of that class.

```python
class Car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def accelerate(self, value):
        self.speed += value
```

`Car()` creates a new object, and each object has its own attributes and methods.

## 2. Encapsulation
Encapsulation hides implementation details and exposes only what is necessary.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # private attribute

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
```

The `__balance` attribute is protected by name mangling, preventing accidental direct access.

## 3. Inheritance
Inheritance allows a class to reuse and extend another class.

```python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"
```

This enables specialization while keeping common behavior in a base class.

## 4. Polymorphism
Polymorphism means different objects can respond to the same method in different ways.

```python
class Cat(Animal):
    def speak(self):
        return "Meow"

animals = [Dog(), Cat()]
for animal in animals:
    print(animal.speak())
```

The same method name (`speak`) behaves differently depending on the object type.

## 5. Abstraction
Abstraction hides complexity and exposes only essential features.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

Abstract classes define a contract, and subclasses must implement required methods.

## 6. Class Methods and Static Methods
- `@staticmethod`: method that does not access instance or class state.
- `@classmethod`: method that works with the class itself.

```python
class MathHelper:
    @staticmethod
    def add(a, b):
        return a + b

    @classmethod
    def description(cls):
        return f"This is {cls.__name__}"
```

## 7. `__init__`, `__str__`, and Other Dunder Methods
Python lets you customize object behavior using special methods.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Person({self.name})"

    def __len__(self):
        return len(self.name)
```

These methods let objects integrate naturally with Python syntax and built-ins.

## 8. Multiple Inheritance and Method Resolution Order (MRO)
A class can inherit from multiple parent classes. Python resolves method lookup using MRO.

```python
class A:
    def hello(self):
        return "A"

class B(A):
    def hello(self):
        return "B"

class C(A):
    def hello(self):
        return "C"

class D(B, C):
    pass
```

Python determines the order in which methods are searched using the MRO.

## 9. Composition over Inheritance
Instead of inheriting everything, it is often better to build classes using smaller objects.

```python
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()

    def drive(self):
        return self.engine.start()
```

This is often cleaner and more flexible than deep inheritance trees.

## 10. Properties and Data Validation
Properties let you manage attribute access in a controlled way.

```python
class Student:
    def __init__(self, age):
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value
```

This improves encapsulation and validation.

## 11. Dataclasses
Dataclasses reduce boilerplate when creating classes mainly for storing data.

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int
```

This automatically generates `__init__`, `__repr__`, and comparison methods.

OOP helps developers:

- build extensible systems
- improve code readability and maintenance
- model real-world entities cleanly
